```python
# FIND-S Algorithm
# Finding the most specific hypothesis from training data

# Training dataset
training_data = [
    ['Sunny', 'Warm', 'Normal', 'Strong', 'Yes'],
    ['Sunny', 'Warm', 'High', 'Strong', 'Yes'],
    ['Rainy', 'Cold', 'High', 'Strong', 'No'],
    ['Sunny', 'Warm', 'High', 'Weak', 'Yes']
]

# Initialize hypothesis with the most specific values
hypothesis = ['Ø'] * (len(training_data[0]) - 1)

print("Initial Hypothesis:", hypothesis)

# FIND-S Algorithm
for example in training_data:

    # Check only positive examples
    if example[-1] == 'Yes':

        for i in range(len(hypothesis)):

            # If hypothesis has no value, assign the example value
            if hypothesis[i] == 'Ø':
                hypothesis[i] = example[i]

            # If values are different, generalize to '?'
            elif hypothesis[i] != example[i]:
                hypothesis[i] = '?'

    print("After processing:", example)
    print("Hypothesis:", hypothesis)

# Final result
print("\nFinal Most Specific Hypothesis:")
print(hypothesis)
```
